```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Testar caminhos para encontrar o ficheiro independentemente de onde o VS Code está a executar
possible_paths = [
    'data/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv',
    '../data/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv',
    r'C:\Users\Guilh\OneDrive\Ambiente de Trabalho\Projetos_IAAC\IAAC_Grupo2\data\Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv'
]

file_path = None
for path in possible_paths:
    if os.path.exists(path):
        file_path = path
        break

if file_path:
    print(f"A carregar o dataset a partir de: {file_path}")
    df = pd.read_csv(file_path)
    
    # Remover espaços extras dos nomes das colunas (ex: ' Label' -> 'Label')
    df.columns = df.columns.str.strip()
    
    print("\n✅ Dataset carregado com sucesso!")
    print(f"Total de Linhas: {df.shape[0]}")
    print(f"Total de Colunas: {df.shape[1]}")
else:
    print("❌ Ficheiro não encontrado. Confirma se o ficheiro está na pasta 'data'.")

# Mostrar as primeiras linhas do dataset
df.head()
```

```python
# 1. Substituir valores infinitos (Inf e -Inf) por NaN
df.replace([np.inf, -np.inf], np.nan, inplace=True)

# 2. Verificar colunas com valores em falta
null_counts = df.isnull().sum()
null_cols = null_counts[null_counts > 0]

print("--- Colunas com valores nulos ou infinitos ---")
if not null_cols.empty:
    print(null_cols)
else:
    print("Nenhum valor em falta encontrado.")
```

```python
# 1. Contagem absoluta e percentagem das classes
print("--- Contagem de Registos por Classe ---")
print(df['Label'].value_counts())

print("\n--- Proporção Percentual (%) ---")
print((df['Label'].value_counts(normalize=True) * 100).round(2))

# 2. Gráfico da distribuição do tráfego
plt.figure(figsize=(8, 4))
sns.countplot(data=df, y='Label', palette='viridis', order=df['Label'].value_counts().index)
plt.title('Distribuição da Variável Target (Label)')
plt.xlabel('Quantidade de Registos')
plt.ylabel('Classe')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.countplot(
    data=df, 
    y='Label', 
    hue='Label', 
    palette='viridis', 
    legend=False, 
    order=df['Label'].value_counts().index
)
plt.title('Distribuição da Variável Target (Label)')
plt.xlabel('Quantidade de Registos')
plt.ylabel('Classe')
plt.tight_layout()
plt.show()
```

```python
# Listar todas as colunas e respetivos tipos de dados
print("--- Informação sobre Colunas e Tipos de Dados ---")
df.info(verbose=True, show_counts=True)

# Identificar colunas de texto / identificadores
text_cols = df.select_dtypes(include=['object']).columns.tolist()
print("\n--- Colunas do tipo Texto / Objeto ---")
print(text_cols)
```

```python
# Versão sem avisos para Pandas 2/3
text_cols = df.select_dtypes(include=['object', 'string']).columns.tolist()
print(text_cols)
```

```python
# 1. Definir e remover colunas de identificação de rede
cols_to_drop = ['Flow ID', 'Source IP', 'Destination IP', 'Timestamp']
existing_cols_to_drop = [col for col in cols_to_drop if col in df.columns]

df_clean = df.drop(columns=existing_cols_to_drop)

# 2. Remover linhas com valores nulos (resultantes da conversão de Inf -> NaN)
df_clean = df_clean.dropna()

# 3. Remover registos duplicados
rows_before = len(df_clean)
df_clean = df_clean.drop_duplicates()
duplicates_removed = rows_before - len(df_clean)

print("✅ Limpeza e filtragem concluídas com sucesso!")
print(f"• Colunas de metadados removidas: {existing_cols_to_drop}")
print(f"• Registos duplicados removidos: {duplicates_removed}")
print(f"• Dimensão final do dataset: {df_clean.shape[0]} linhas e {df_clean.shape[1]} colunas")
```

```python
# Mapear BENIGN para 0 e DDoS para 1
label_map = {'BENIGN': 0, 'DDoS': 1}
df_clean['Label_Num'] = df_clean['Label'].map(label_map)

print("--- Confirmação do Mapeamento ---")
print(df_clean[['Label', 'Label_Num']].value_counts())
```

```python
# Filtrar colunas numéricas para o cálculo de correlação
numeric_df = df_clean.select_dtypes(include=['number'])

# Calcular a correlação de todas as variáveis com a classe de ataque
correlations = numeric_df.corr()['Label_Num'].sort_values(ascending=False)

print("--- Top 10 Features Mais Correlacionadas com DDoS (Positiva) ---")
print(correlations.head(11).tail(10))  # Exclui a própria variável Label_Num

print("\n--- Top 10 Features Com Maior Correlação Negativa ---")
print(correlations.tail(10))
```

```python
# 1. Identificar e remover colunas com variância zero (constantes)
constant_cols = [col for col in numeric_df.columns if numeric_df[col].std() == 0]

# 2. Definir portas de rede para remoção (se ainda estiverem no dataset)
port_cols = ['Source Port', 'Destination Port']
extra_cols_to_drop = constant_cols + [col for col in port_cols if col in df_clean.columns]

# 3. Criar dataset final da EDA
df_eda = df_clean.drop(columns=extra_cols_to_drop)

print(f"✅ Remoção concluída!")
print(f"• Colunas constantes removidas ({len(constant_cols)}): {constant_cols}")
print(f"• Colunas de portas removidas: {[col for col in port_cols if col in df_clean.columns]}")
print(f"• Dimensão final para treino: {df_eda.shape[0]} linhas e {df_eda.shape[1]} colunas")
```

```python
# Recalcular as top 10 features após a limpeza
top_10_features = correlations.head(11).tail(10).index.tolist()

# Criar matriz de correlação reduzida
plt.figure(figsize=(10, 8))
sns.heatmap(
    df_eda[top_10_features + ['Label_Num']].corr(), 
    annot=True, 
    fmt='.2f', 
    cmap='coolwarm', 
    cbar=True
)
plt.title('Heatmap: Top 10 Features Correlacionadas com DDoS')
plt.tight_layout()
plt.show()
```

```python
# 1. Recalcular as correlações usando apenas o dataset limpo (df_eda)
eda_numeric = df_eda.select_dtypes(include=['number'])
eda_correlations = eda_numeric.corr()['Label_Num'].sort_values(ascending=False)

# 2. Selecionar as top 10 features (excluindo a própria Label_Num)
top_10_features = eda_correlations.head(11).tail(10).index.tolist()

# 3. Gerar o Heatmap de correlação
plt.figure(figsize=(10, 8))
sns.heatmap(
    df_eda[top_10_features + ['Label_Num']].corr(), 
    annot=True, 
    fmt='.2f', 
    cmap='coolwarm', 
    cbar=True
)
plt.title('Heatmap: Top 10 Features Correlacionadas com DDoS')
plt.tight_layout()
plt.show()
```
